```python
var('x,y,z,v_x,v_y,v_z,rho,t')

variables = [x, y, z, v_x, v_y, v_z, rho]

rhs = [
    v_x,
    v_y,
    v_z,
    -x*rho^3 + 3*x + 2*v_y,
    -y*rho^3 - 2*v_x,
    -z*rho^3 - z,
    -rho^3*(x*v_x + y*v_y + z*v_z),
]
```

```python
initial_conditions = [1, 0, 0, 0, 0, 0, 1]

problem = Initial_problem(
    variables,
    rhs,
    initial_conditions,
    1,
)
```

```python
quad_problem, quad_system = problem.quadratize(
    method='qbee',
    time_var=t,
    allow_laurent=False,
    max_new_vars=30,
    max_nodes=5000,
    beam_width=40,
)
```

```python
assert quad_system.is_quadratic()
assert len(quad_problem.x) == len(quad_problem.f)
assert len(quad_problem.x) == len(quad_problem.x0)

darboux = (x^2 + y^2 + z^2)*rho^2 - 1
initial_substitution = [
    variable == value
    for variable, value in zip(variables, initial_conditions)
]
assert darboux.subs(initial_substitution) == 0

print('Квадратизация системы Хилла успешно завершена.')
print('Исходных переменных:', len(variables))
print('Созданных переменных:', len(quad_system.generated_variables()))
print('Всего переменных после квадратизации:', len(quad_system.state_variables()))
```
